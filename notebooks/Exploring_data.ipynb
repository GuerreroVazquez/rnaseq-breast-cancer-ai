```python
import pandas as pd
```

# Lokking up the data

```python
X_train_all = pd.read_csv('X_train_all.csv')
```

```python
X_train_all.median(axis=1).abs().max()
```

```python
import matplotlib.pyplot as plt

plt.hist(X_train_all.median(axis=1), bins=50)
plt.title("Distribution of gene-wise medians")
plt.xlabel("Median value")
plt.ylabel("Number of genes")
plt.show()
```

```python
# Get the median of each column
X_train_median = X_train_all.median()
X_train_median
```

```python
# Get the median of each row
X_train_median_row = X_train_all.median(axis=1)
X_train_median_row
```

```python
import seaborn as sns
# plot kernel density estimate of the median of random 100 rows
sns.kdeplot(X_train_median_row.sample(100), fill=True)
import matplotlib.pyplot as plt
plt.title('Kernel Density Estimate of Row Medians')
plt.xlabel('Row Median Value')
plt.ylabel('Density')
plt.show()
# Get the median of each column
X_train_median = X_train_all.median()
# Get the median of each row
X_train_median_row = X_train_all.median(axis=1)
# Plot the kernel density estimate of the median of random 100 rows
sns.kdeplot(X_train_median_row.sample(100), fill=True)
```

```python
from scipy.stats import shapiro
stat, p = shapiro(X_train_median_row.sample(100))
print(f'Shapiro-Wilk test: W={stat:.3f}, p-value={p:.3g}')
```
