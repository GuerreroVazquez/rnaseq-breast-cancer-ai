```python
from scipy.stats import chi2_contingency
sifnificant_correlation = []
significant_correlation_columns=[]
def chi2_test(data, col1, col2):
    """Performs a Chi-squared test of independence between two categorical columns."""
    contingency_table = pd.crosstab(data[col1], data[col2])
    chi2, p, dof, expected = chi2_contingency(contingency_table)
    if p < 0.05:
        print(f"Significant association found between {col1} and {col2}")
        sifnificant_correlation.append((col1, col2, chi2, p))
        significant_correlation_columns.append(col1)

for col_1 in missing_values.index:
  for col_2 in missing_values.index:
    if col_1 != col_2:
      chi2_test(expression_df, col_1, col_2)
```

```python
def analyze_convergence(df, biomarker):
    """Analyzes the convergence between MGC and SGC predictions for a given biomarker."""
    mgc_col = f'{biomarker} prediction mgc'
    sgc_col = f'{biomarker} prediction sgc'

    # Drop rows where MGC or SGC values are NaN
    df = df.dropna(subset=[mgc_col, sgc_col])

    # Convert to numeric, coerce errors to NaN
    df[mgc_col] = pd.to_numeric(df[mgc_col], errors='coerce')
    df[sgc_col] = pd.to_numeric(df[sgc_col], errors='coerce')

    # Drop rows where conversion to numeric failed
    df = df.dropna(subset=[mgc_col, sgc_col])

    # Convert to int
    df[mgc_col] = df[mgc_col].astype(int)
    df[sgc_col] = df[sgc_col].astype(int)

    # Calculate Agreement:
    agreement = (df[mgc_col] == df[sgc_col]).mean()
    print(f"Agreement between MGC and SGC for {biomarker}: {agreement:.2f}")

    #Calculate a contingency table
    contingency_table = pd.crosstab(df[mgc_col], df[sgc_col])
    print ("\nContingency Table:")
    print(contingency_table)
    return df


biomarkers = ['er', 'pgr', 'her2', 'ki67']
converged_missing_status = {}
for biomarker in biomarkers:
    print(f"\nAnalyzing {biomarker.upper()} status predictions:")
    converged_missing_status[biomarker] = analyze_convergence(missing_status.copy(), biomarker)  # Pass a copy to avoid modifying original df


# 2. Replace Original Values with Convergent Predictions:

def replace_with_convergent(main_df, converged_df, biomarker):
    """Replaces original missing values with convergent MGC/SGC predictions."""
    mgc_col = f'{biomarker} prediction mgc'
    sgc_col = f'{biomarker} prediction sgc'
    original_col = f'{biomarker} status'

    for index, row in converged_df.iterrows():
        if row[mgc_col] == row[sgc_col]:  # Predictions converge
            main_df.loc[index, original_col] = row[mgc_col]  # Replace original value
            #main_df.loc[index, original_col] = int(row[mgc_col])  # Replace original value
    return main_df

for biomarker in biomarkers:
    smal_expression_conveted_df = replace_with_convergent(smal_expression_df, converged_missing_status[biomarker], biomarker)

smal_expression_conveted_df
```

```python
def generate_missing_value_report(df):
    """
    Generates a report of missing values and their co-occurrences in a DataFrame.

    Args:
        df (pd.DataFrame): The input DataFrame.

    Returns:
        None: Prints the report to the console.
    """

    total_rows = len(df)
    missing_report = {}

    # Calculate missing value counts for each column
    for col in df.columns:
        missing_count = df[col].isnull().sum()
        missing_percentage = (missing_count / total_rows) * 100
        missing_report[col] = {'missing_count': missing_count, 'missing_percentage': missing_percentage}

    # Sort columns by missing percentage in descending order
    sorted_cols = sorted(missing_report.keys(), key=lambda k: missing_report[k]['missing_percentage'], reverse=True)

    print("Missing Value Report:")
    for col in sorted_cols:
        print(f"{col}: Missing {missing_report[col]['missing_count']} values ({missing_report[col]['missing_percentage']:.2f}%)")

    # Calculate co-occurrence of missing values
    missing_value_indicator_df = df.isnull().astype(int)
    co_occurrence_matrix = missing_value_indicator_df.T.dot(missing_value_indicator_df)

    return co_occurrence_matrix

# Assuming 'smal_expression_df' is your DataFrame:
generate_missing_value_report(smal_expression_df)
```
