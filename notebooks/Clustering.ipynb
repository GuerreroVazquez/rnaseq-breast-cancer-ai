```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.decomposition import PCA
from sklearn.cluster import SpectralClustering, KMeans
import umap

def visualize_clusters(data: pd.DataFrame, pam50: pd.Series, title_prefix: str = "", n_components_pca: int = 2, n_clusters: int = 5, umap_n_neighbors: int = 15, umap_min_dist: float = 0.1):
    """
    Visualizes clusters in the data using PCA, Spectral Clustering, UMAP, and K-means,
    colored by PAM50 subtype.

    Args:
        data: DataFrame of data (e.g., gene expression, encoded features).
        pam50: Series of PAM50 subtypes.
        title_prefix: Prefix for plot titles.
        n_components_pca: Number of PCA components to retain.
        n_clusters: Number of clusters for Spectral Clustering and K-means.
        umap_n_neighbors: UMAP parameter (number of neighbors).
        umap_min_dist: UMAP parameter (minimum distance).
    """

    # 1. Preprocessing
    # Scale the data
    scaler = StandardScaler()
    data_scaled = scaler.fit_transform(data)

    print("Data scaled successfully.")

    # Encode PAM50 subtypes (for coloring)
    label_encoder = LabelEncoder()
    pam50_encoded = label_encoder.fit_transform(pam50)
    pam50_labels = label_encoder.classes_  # Store the original class labels

    #print("PAM50 subtypes encoded successfully.")

    # 2. PCA
    pca = PCA(n_components=n_components_pca)
    data_pca = pca.fit_transform(data_scaled)
    #print("PCA completed successfully.")


    # 3. UMAP
    #reducer = umap.UMAP(n_neighbors=umap_n_neighbors, min_dist=umap_min_dist, random_state=42)
    #data_umap = reducer.fit_transform(data_scaled)
    #print("UMAP completed successfully.")


    # 4. Spectral Clustering
    #spectral_clustering = SpectralClustering(n_clusters=n_clusters, assign_labels='discretize', random_state=42)
    #spectral_labels = spectral_clustering.fit_predict(data_scaled)
    #print("Spectral Clustering completed successfully.")


    # 5. K-means
    kmeans = KMeans(n_clusters=n_clusters, random_state=42, n_init=10)  # Explicitly set n_init
    kmeans_labels = kmeans.fit_predict(data_scaled)
    print("K-means clustering completed successfully.")

    # 6. Plotting Function (to avoid repetition)
    def plot_clusters(data_reduced, labels, title, xtitle, ytitle, pam50_labels):
        """Plots clusters, colored by PAM50 subtypes."""
        plt.figure(figsize=(8, 6))
        scatter = plt.scatter(data_reduced[:, 0], data_reduced[:, 1], c=labels, cmap='viridis', s=50)  # Use viridis colormap
        plt.title(title)
        plt.xlabel(xtitle)
        plt.ylabel(ytitle)

        # Add legend for PAM50 subtypes (using the scatter plot)
        legend_labels = [pam50_labels[i] for i in np.unique(labels)] # Get unique PAM50 class labels
        plt.legend(handles=scatter.legend_elements()[0], labels=legend_labels, title="PAM50 Subtype") # Add legend

        plt.colorbar(scatter, label="Cluster") # Adds the colorbar
        plt.grid(True)
        plt.savefig(f"{title}.svg")  # Save the plot
        plt.show()

    # 7. Plotting (using the function)
    #plot_clusters(data_pca, pam50_encoded, f"{title_prefix}PCA Colored by PAM50", "PCA Component 1", "PCA Component 2", pam50_labels)
    #print("PCA plot completed successfully.")
    #plot_clusters(data_umap, pam50_encoded, f"{title_prefix}UMAP Colored by PAM50", "UMAP Dimension 1", "UMAP Dimension 2", pam50_labels)
    #print("UMAP plot completed successfully.")

    # Separate plotting for spectral and kmeans
    #plot_clusters(data_pca, spectral_labels, f"{title_prefix}Spectral Clustering Colored by Spectral Cluster", "PCA Component 1", "PCA Component 2", pam50_labels)
    #print("Spectral Clustering plot completed successfully.")
    plot_clusters(data_pca, kmeans_labels, f"{title_prefix}K-means Clustering Colored by K-Means Cluster", "PCA Component 1", "PCA Component 2", pam50_labels)
    print("K-means plot completed successfully.")

    print ("Visualisation ready! You can view the plots.")
```

```python
pam50_data = clinical_train_df['pam50 subtype'].astype(str)  # Ensure PAM50 is a string type for consistency
visualize_clusters(X_train_encoded_df, pam50_data, title_prefix="Autoencoder ", n_clusters=5)
visualize_clusters(X_train_selected_df, pam50_data, title_prefix="Selected Genes ", n_clusters=5)
```

```python
pam50_data = clinical_train_df['pam50 subtype'].astype(str)  # Ensure PAM50 is a string type for consistency
visualize_clusters(X_train_all, pam50_data, title_prefix="Original ", n_clusters=5)
```
